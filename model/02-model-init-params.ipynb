```python
import os
from dotenv import load_dotenv
# 加载配置文件
load_dotenv(override=True)
from langchain.chat_models import init_chat_model

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(
    model="openai:deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    temperature=0,
)


print(model.invoke("张三，男，30岁，拥有八年编程开发经验，目前在互联网某大厂担任技术专家。帮我从上文中提取数据，返回Json格式。").content)
```

## Token

不同模型厂家使用不同的分词算法，会导致同一段文本在不同的模型中Token数量不同，但大约可以进行估计。

- 一个中文Token大约是1-1.8个汉字，一个英文Token大约是3-4个字符。

```python
model = init_chat_model(
    model="openai:deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    # temperature=2,
    max_tokens=3,
)

print(model.invoke("介绍一下你自己").content)
```
